### Policy based

```python
def cal_r(policy, reward):
    state = ['E', 'G', 'B']
    r = np.zeros((3,1))
    for i, action in enumerate(policy):
        key = state[i] + action
        r[i,0] = reward[key]
    return r

def cal_P(policy,P):
    state = ['E', 'G', 'B']
    P_MAT = np.zeros((3,3))
    for i, action in enumerate(policy):
        for j, next_state in enumerate(state):
            key = state[i] + action + next_state
            P_MAT[i,j] = P[key]
    return P_MAT

def cal_V(P, r, gamma):
    I = np.eye(3)
    V = np.linalg.solve(I - gamma * P, r)
    return V

def opt_p(V, reward, P, gamma):
    state = ['E', 'G', 'B']
    actions = ['F', 'S']
    policy = []
    for i in range(3):
        best_action = None
        best_value = -np.inf
        for action in actions:
            key = state[i] + action
            r = reward[key]
            P_row = np.array([P[state[i] + action + next_state] for next_state in state])
            value = r + gamma * np.dot(P_row, V)
            if value > best_value:
                best_value = value
                best_action = action
        policy.append(best_action)
    return policy
```

```python
policy_1 = ['F', 'F', 'F']
policy_2 = ['S', 'F', 'F']
while policy_1 != policy_2:
    policy_1 = policy_2
    r = cal_r(policy_1, reward)
    P_matrix = cal_P(policy_1, P)
    V = cal_V(P_matrix, r, gamma)
    policy_2 = opt_p(V, reward, P, gamma)
```

```python
V = cal_V(cal_P(policy_2, P), cal_r(policy_2, reward), gamma)
Policy = opt_p(V, reward, P, gamma)
print("Optimal Policy:", Policy)
print("Value Function:", V)
```

### 修正策略迭代

```python
import numpy as np

# =========================
# 1. 参数设置
# =========================

states = ['E', 'G', 'B']
actions = ['F', 'S']

gamma = 0.9

reward = {
    'EF': -90, 'GF': -90, 'BF': -90,
    'ES': 100, 'GS': 70, 'BS': 20
}

P = {
    'EFE': 1, 'EFG': 0, 'EFB': 0,
    'GFE': 1, 'GFG': 0, 'GFB': 0,
    'BFE': 1, 'BFG': 0, 'BFB': 0,

    'ESE': 0.7, 'ESG': 0.2, 'ESB': 0.1,
    'GSE': 0, 'GSG': 0.7, 'GSB': 0.3,
    'BSE': 0, 'BSG': 0.1, 'BSB': 0.9
}

# =========================
# 2. 初始化价值函数
# =========================

V = np.zeros(3)

print("初始价值函数 V0:")
print(V)

# =========================
# 3. 执行两次值迭代
# =========================

for k in range(100):

    V_new = np.zeros(3)

    policy = []

    for i, s in enumerate(states):

        Q_values = {}

        for a in actions:

            r = reward[s + a]

            P_row = np.array([
                P[s + a + next_s] for next_s in states
            ])

            Q = r + gamma * np.dot(P_row, V)

            Q_values[a] = Q

        # 选择最大 Q 值对应的动作
        best_action = max(Q_values, key=Q_values.get)

        V_new[i] = max(Q_values.values())

        policy.append(best_action)

    V = V_new

    print(f"\n第 {k+1} 次迭代:")
    print("价值函数:", V)
    print("当前贪心策略:", policy)
```

### Linear Programming

```python
import cvxpy as cp

gamma = 0.9

states = ['E', 'G', 'B']
actions = ['F', 'S']

reward = {
    'EF': -90, 'GF': -90, 'BF': -90,
    'ES': 100, 'GS': 70, 'BS': 20
}

P = {
    'EFE': 1, 'EFG': 0, 'EFB': 0,
    'GFE': 1, 'GFG': 0, 'GFB': 0,
    'BFE': 1, 'BFG': 0, 'BFB': 0,

    'ESE': 0.7, 'ESG': 0.2, 'ESB': 0.1,
    'GSE': 0, 'GSG': 0.7, 'GSB': 0.3,
    'BSE': 0, 'BSG': 0.1, 'BSB': 0.9
}


V = cp.Variable(3)

constraints = []

for i, s in enumerate(states):
    for a in actions:

        r = reward[s + a]

        P_row = np.array([
            P[s + a + next_s] for next_s in states
        ])

        constraints.append(
            V[i] >= r + gamma * P_row @ V
        )


objective = cp.Minimize(cp.sum(V))

problem = cp.Problem(objective, constraints)

problem.solve()


print("求解状态:", problem.status)
print("最优目标函数值:", problem.value)
print("最优价值函数:")

for i, s in enumerate(states):
    print(f"V({s}) = {V.value[i]:.6f}")

print("\n最优策略:")

for i, s in enumerate(states):

    Q_values = {}

    for a in actions:
        r = reward[s + a]

        P_row = np.array([
            P[s + a + next_s] for next_s in states
        ])

        Q_values[a] = r + gamma * P_row @ V.value

    best_action = max(Q_values, key=Q_values.get)

    print(
        f"状态 {s}: "
        f"Q(F)={Q_values['F']:.6f}, "
        f"Q(S)={Q_values['S']:.6f}, "
        f"选择 {best_action}"
    )
```
